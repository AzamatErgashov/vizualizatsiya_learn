```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
import numpy as np 
```

```python
df = pd.read_csv("mers.csv", index_col=0)
df=df[df.year>=2010] # 2010-yildan keyin chiqqan mashinalarni saqlab qolamiz
df=df[df.transmission!='Other'] # Faqat avtomat, mexanik va yarim-avtomat mashinalarni olib qolamiz
df=df[df.fuelType!='Other'] # Faqat benzin, dizel va gibrid mashinalarni olib qolamiz
df.head()
```

```python
plt.figure(figsize=(16,6))# chizmaga olcham berish
plt.subplot(2,1,1)
x = np.arange(0, 4*np.pi, 0.1)
sin = np.sin(x)
cos = np.cos(x)


plt.plot(x,sin,
        linewidth=1,
        linestyle="-",
        color="lime", 
        alpha=0.7,
        marker="o")
plt.grid(color='k', linestyle='--')
plt.subplot(2,1,2)

plt.plot(x, cos,
        linewidth=1,
        linestyle="-",
        color="red", 
        alpha=0.7,
        marker="o")

plt.title('sin(x) va cos(x)')
plt.legend(['sin(x)','cos(x)'], loc=0, fontsize='x-large')
plt.xlabel("pi")# x uqga label quyish
plt.ylabel('data')# y uqga label quysih
plt.grid(color='k', linestyle='--')# chziq chizib beradi
plt.show()
```

```python
sns.scatterplot(data=df, x='price', y='year')
plt.show()
```

```python
fig, ax = plt.subplots(2,1, figsize=(13,6))
sns.histplot(ax=ax[0], data=df, x='price')
sns.scatterplot(ax=ax[1], data=df, x='price', y='year')
plt.show()
```

```python
fig, ax = plt.subplots(2,2, figsize=(16,10))
fig.suptitle('grafiklar')
sns.histplot(ax=ax[0,0], data=df, x='price')
sns.histplot(ax=ax[0,1], data=df, x='year')
sns.scatterplot(ax=ax[1,0], data=df, x='year',y='price', hue='fuelType', s=50)
sns.scatterplot(ax=ax[1,1], data=df, x='year', y='tax',hue='fuelType', s=50)
ax[0,0].set_title('price')
ax[0,1].set_title('year')
ax[1,0].set_title('year + price')
ax[1,1].set_title('tax + year')

plt.show()
```
