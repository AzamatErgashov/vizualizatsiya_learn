```python
x = np.arange(0, 4*np.pi, 0.1)
y = np.sin(x)
plt.plot(x,y)
plt.show()
```

# linewidth - chiziqni qalinligi 
# linestyle - chiziq korinishi
# color - rang beradi
# marker - orasidan uzgartirish

```python
plt.plot(x,y, linewidth=1, linestyle="-", color="lime", marker="o")
plt.show()
```

```python
plt.plot(x,y, linewidth=1, linestyle="-", color="lime", marker="o")
plt.xlabel("pi")# x uqga label quyish
plt.ylabel('data')# y uqga label quysih
plt.grid(color='k', linestyle='--')# chziq chizib beradi
plt.show()
```

```python
plt.figure(figsize=(6,3))# chizmaga olcham berish
plt.plot(x,y, linewidth=1, linestyle="-", color="lime", marker="o")
plt.xlabel("pi")# x uqga label quyish
plt.ylabel('data')# y uqga label quysih
plt.grid(color='k', linestyle='--')# chziq chizib beradi
plt.show()
```

# plt.legend bizga chizmamizda ichida aloxida tortburchak yaratib label berishga yozdam beradi
* loc funksiyasi esa usha turtburchak qaysi joyda turishini aytadi

```python
plt.figure(figsize=(16,6))# chizmaga olcham berish
x = np.arange(0, 4*np.pi, 0.1)
sin = np.sin(x)
cos = np.cos(x)


plt.plot(x,sin,
        linewidth=1,
        linestyle="-",
        color="lime", 
        alpha=0.7,
        marker="o")

plt.plot(x, cos,
        linewidth=1,
        linestyle="-",
        color="red", 
        alpha=0.7,
        marker="o")

plt.title('sin(x) va cos(x)')
plt.legend(['sin(x)','cos(x)'], loc=0, fontsize='x-large')
plt.xlabel("pi")# x uqga label quyish
plt.ylabel('data')# y uqga label quysih
plt.grid(color='k', linestyle='--')# chziq chizib beradi
plt.savefig('vizual.jpg')
plt.show()
```
