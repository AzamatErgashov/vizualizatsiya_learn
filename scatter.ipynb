```python
import pandas as pd 
import numpy as np 
import seaborn as sns 
import matplotlib.pyplot as plt 
```

```python
df = pd.read_csv("mers.csv", index_col=0)
df.head()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='year', y='price')
plt.show()
```

* bu yerda kakkalanib qolgan qiymatlarni olib tashlashimiz kerak

```python
df = df[df['year']!=1970]# bu yerda 1970 yildagi moshinlar bizga kerakmas deb aytyapmiz
plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='year', y='price')
plt.show()
```

# mana endi datamiz osayaptimi yoqmi buni kordik amma chizmada korsatish uchun regplotdan foydalanimiz

```python
plt.figure(figsize=(12,6))
df = df[df['year'] > 2000]
sns.regplot(data=df, x='year', y='price')
plt.show()
```

```python
df = df[df['year']!=1970]# bu yerda 1970 yildagi moshinlar bizga kerakmas deb aytyapmiz
plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='mileage', y='price')
plt.show()
```

* bu yerda esa moshinlarni kilometri oshgan sari narxi tushub bormoqda

# s= bu dumaloqlarning size degani

```python
df = df[df['year']!=1970]# bu yerda 1970 yildagi moshinlar bizga kerakmas deb aytyapmiz
plt.figure(figsize=(16,6))
sns.scatterplot(data=df, x='mileage', y='price', hue='transmission', style='fuelType', s=100)
plt.show()
```

# endi korishimiz mumkin style orqali chizmaga yana bir parametr qushishimiz mumkin ekan

```python
sns.swarmplot(data=df, x=df['fuelType'], y=df['price'])
plt.show()
```

```python
sns.stripplot(data=df, x=df['fuelType'], y=df['price'])
plt.show()
```
